```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr

from google.colab import files
uploaded = files.upload()

image_path = list(uploaded.keys())[0]

img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

print("Image Shape:", img.shape)
noise = np.random.normal(0,25,gray.shape)

noisy = gray + noise

noisy = np.clip(noisy,0,255).astype(np.uint8)
gaussian = cv2.GaussianBlur(noisy,(5,5),0)
median = cv2.medianBlur(noisy,5)
normalized = cv2.normalize(
    median,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)
titles = [
    "Original",
    "Noisy",
    "Gaussian Blur",
    "Median Blur",
    "Normalized"
]

images = [
    gray,
    noisy,
    gaussian,
    median,
    normalized
]

plt.figure(figsize=(18,6))

for i in range(len(images)):
    plt.subplot(1,5,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis("off")

plt.show()
orb = cv2.ORB_create(nfeatures=500)

kp1, des1 = orb.detectAndCompute(gray,None)

kp2, des2 = orb.detectAndCompute(noisy,None)

kp3, des3 = orb.detectAndCompute(gaussian,None)

kp4, des4 = orb.detectAndCompute(median,None)

kp5, des5 = orb.detectAndCompute(normalized,None)

counts = [
    len(kp1),
    len(kp2),
    len(kp3),
    len(kp4),
    len(kp5)
]

print(counts)
draw1 = cv2.drawKeypoints(gray,kp1,None,color=(0,255,0))

draw2 = cv2.drawKeypoints(noisy,kp2,None,color=(0,255,0))

draw3 = cv2.drawKeypoints(gaussian,kp3,None,color=(0,255,0))

draw4 = cv2.drawKeypoints(median,kp4,None,color=(0,255,0))

draw5 = cv2.drawKeypoints(normalized,kp5,None,color=(0,255,0))
imgs = [
    draw1,
    draw2,
    draw3,
    draw4,
    draw5
]

plt.figure(figsize=(18,6))

for i in range(5):
    plt.subplot(1,5,i+1)
    plt.imshow(imgs[i])
    plt.title(titles[i])
    plt.axis("off")

plt.show()
methods = [
    gray,
    noisy,
    gaussian,
    median,
    normalized
]

psnr_values = []

ssim_values = []

for m in methods:

    p = psnr(gray,m,data_range=255)

    s = ssim(gray,m,data_range=255)

    psnr_values.append(round(p,2))

    ssim_values.append(round(s,3))

print(psnr_values)

print(ssim_values)
table = pd.DataFrame({

    "Method":titles,

    "Keypoints":counts,

    "PSNR":psnr_values,

    "SSIM":ssim_values

})

table
plt.figure(figsize=(8,5))

plt.bar(table["Method"],table["Keypoints"])

plt.title("Feature Detection Comparison")

plt.ylabel("Detected Keypoints")

plt.xticks(rotation=20)

plt.show()
plt.figure(figsize=(8,5))

plt.plot(table["Method"],table["PSNR"],marker='o')

plt.title("PSNR Comparison")

plt.ylabel("PSNR")

plt.grid()

plt.show()
plt.figure(figsize=(8,5))

plt.plot(table["Method"],table["SSIM"],marker='o')

plt.title("SSIM Comparison")

plt.ylabel("SSIM")

plt.grid()

plt.show()
plt.figure(figsize=(10,5))

x = np.arange(len(titles))

width = 0.3

plt.bar(x-width,table["PSNR"],width,label="PSNR")

plt.bar(x,table["SSIM"]*100,width,label="SSIM ×100")

plt.bar(x+width,table["Keypoints"],width,label="Keypoints")

plt.xticks(x,titles,rotation=20)

plt.legend()

plt.title("Overall Comparison")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from google.colab import files
uploaded = files.upload()

image_path = list(uploaded.keys())[0]

image = cv2.imread(image_path)

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

print("Image Shape:", image.shape)
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
threshold = 127

_, binary = cv2.threshold(gray,
                          threshold,
                          255,
                          cv2.THRESH_BINARY)
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("RGB Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray,cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(binary,cmap='gray')
plt.title("Binary Image")
plt.axis("off")

plt.show()
edges_gray = cv2.Canny(gray,100,200)

edges_binary = cv2.Canny(binary,100,200)
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(edges_gray,cmap='gray')
plt.title("Edges from Grayscale")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(edges_binary,cmap='gray')
plt.title("Edges from Binary")
plt.axis("off")

plt.show()
orb = cv2.ORB_create(nfeatures=500)

kp_gray, des_gray = orb.detectAndCompute(gray,None)

kp_binary, des_binary = orb.detectAndCompute(binary,None)

print("Gray Keypoints:",len(kp_gray))

print("Binary Keypoints:",len(kp_binary))
gray_features = cv2.drawKeypoints(gray,
                                  kp_gray,
                                  None,
                                  color=(0,255,0))

binary_features = cv2.drawKeypoints(binary,
                                    kp_binary,
                                    None,
                                    color=(0,255,0))
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(gray_features)
plt.title("ORB Features (Grayscale)")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary_features)
plt.title("ORB Features (Binary)")
plt.axis("off")

plt.show()
gray_edges = np.count_nonzero(edges_gray)

binary_edges = np.count_nonzero(edges_binary)

print("Edge Pixels (Gray):",gray_edges)

print("Edge Pixels (Binary):",binary_edges)
table = pd.DataFrame({

    "Representation":[
        "Grayscale",
        "Binary"
    ],

    "Edge Pixels":[
        gray_edges,
        binary_edges
    ],

    "ORB Keypoints":[
        len(kp_gray),
        len(kp_binary)
    ]

})

table
plt.figure(figsize=(7,5))

plt.bar(table["Representation"],
        table["Edge Pixels"])

plt.title("Edge Detection Comparison")

plt.ylabel("Edge Pixels")

plt.show()
plt.figure(figsize=(7,5))

plt.bar(table["Representation"],
        table["ORB Keypoints"])

plt.title("Feature Extraction Comparison")

plt.ylabel("Detected Keypoints")

plt.show()
plt.figure(figsize=(8,5))

x = np.arange(2)

width = 0.35

plt.bar(x-width/2,
        table["Edge Pixels"],
        width,
        label="Edges")

plt.bar(x+width/2,
        table["ORB Keypoints"],
        width,
        label="Keypoints")

plt.xticks(x,
           table["Representation"])

plt.legend()

plt.title("Image Representation Comparison")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from google.colab import files
uploaded = files.upload()
image_path = list(uploaded.keys())[0]
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
print("Image Shape:", image.shape)
sobel_x = cv2.Sobel(gray,
                    cv2.CV_64F,
                    1,
                    0,
                    ksize=3)
sobel_y = cv2.Sobel(gray,
                    cv2.CV_64F,
                    0,
                    1,
                    ksize=3)
sobel = cv2.magnitude(sobel_x, sobel_y)
sobel = np.uint8(np.clip(sobel, 0, 255))
canny = cv2.Canny(gray,
                  100,
                  200)
plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(sobel, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(canny, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
sobel_edges = np.count_nonzero(sobel)
canny_edges = np.count_nonzero(canny)
print("Sobel Edge Pixels :", sobel_edges)
print("Canny Edge Pixels :", canny_edges)
orb = cv2.ORB_create(nfeatures=500)
kp_sobel, des_sobel = orb.detectAndCompute(sobel, None)
kp_canny, des_canny = orb.detectAndCompute(canny, None)
print("ORB Features on Sobel :", len(kp_sobel))
print("ORB Features on Canny :", len(kp_canny))
sobel_features = cv2.drawKeypoints(
    sobel,
    kp_sobel,
    None,
    color=(0,255,0),
    flags=0
)
canny_features = cv2.drawKeypoints(
    canny,
    kp_canny,
    None,
    color=(0,255,0),
    flags=0
)
plt.figure(figsize=(14,6))
plt.subplot(1,2,1)
plt.imshow(sobel_features)
plt.title("ORB Features - Sobel")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(canny_features)
plt.title("ORB Features - Canny")
plt.axis("off")
plt.show()
results = pd.DataFrame({
    "Method":[
        "Sobel",
        "Canny"
    ],
    "Edge Pixels":[
        sobel_edges,
        canny_edges
    ],
    "ORB Features":[
        len(kp_sobel),
        len(kp_canny)
    ]
})
print("\nSummary Table\n")
print(results)
plt.figure(figsize=(7,5))
plt.bar(results["Method"],
        results["Edge Pixels"])
plt.title("Edge Pixels Comparison")
plt.xlabel("Method")
plt.ylabel("Number of Edge Pixels")
plt.grid(axis='y')
plt.show()
plt.figure(figsize=(7,5))
plt.bar(results["Method"],
        results["ORB Features"])
plt.title("ORB Feature Comparison")
plt.xlabel("Method")
plt.ylabel("Number of Features")
plt.grid(axis='y')
plt.show()
x = np.arange(len(results))
width = 0.35
plt.figure(figsize=(8,5))
plt.bar(x-width/2,
        results["Edge Pixels"],
        width,
        label="Edge Pixels")
plt.bar(x+width/2,
        results["ORB Features"],
        width,
        label="ORB Features")
plt.xticks(x,
           results["Method"])
plt.title("Sobel vs Canny Comparison")
plt.xlabel("Edge Detection Method")
plt.ylabel("Count")
plt.legend()
plt.grid(axis='y')
plt.show()
results.to_csv("Edge_Detection_Comparison.csv", index=False)
print("\nResults saved as Edge_Detection_Comparison.csv")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files

uploaded = files.upload()

img = cv2.imread(next(iter(uploaded)))

rows, cols = img.shape[:2]

src_points = np.float32([
    [0, 0],
    [cols - 1, 0],
    [0, rows - 1],
    [cols - 1, rows - 1]
])

dst_points = np.float32([
    [0, 0],
    [cols - 1, 0],
    [int(0.33 * cols), rows - 1],
    [int(0.66 * cols), rows - 1]
])

M = cv2.getPerspectiveTransform(src_points, dst_points)

perspective_img = cv2.warpPerspective(img, M, (cols, rows))

print("Original Image:")
cv2_imshow(img)

print("Perspective Transformed Image:")
cv2_imshow(perspective_img)
cv2.imwrite("Perspective_Transformed_Image.jpg", perspective_img)
```

```python
import cv2
import numpy as np
from google.colab import files

uploaded = files.upload()

roi_points = np.array([
    (150, 200),
    (450, 200),
    (550, 500),
    (50, 500)
])

target_points = np.array([
    (0, 0),
    (400, 0),
    (400, 600),
    (0, 600)
])

M = cv2.getPerspectiveTransform(
    roi_points.astype(np.float32),
    target_points.astype(np.float32)
)

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    dst = cv2.warpPerspective(frame, M, (400, 600))

    cv2.imshow("Original Frame", frame)
    cv2.imshow("Transformed Frame", dst)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

uploaded = files.upload()

img = cv2.imread(next(iter(uploaded)))

rows, cols = img.shape[:2]

src_points = np.float32([
    [0, 0],
    [cols - 1, 0],
    [0, rows - 1],
    [cols - 1, rows - 1]
])

dst_points = np.float32([
    [0, 0],
    [cols - 1, 0],
    [0, int(0.7 * rows)],
    [cols - 1, int(0.7 * rows)]
])

M, _ = cv2.findHomography(src_points, dst_points)

homography_img = cv2.warpPerspective(
    img, M, (cols, rows)
)

cv2_imshow(img)

cv2_imshow(homography_img)

cv2.imwrite(
    "transformation_using_Homography_Image.jpg",
    homography_img
)
```
